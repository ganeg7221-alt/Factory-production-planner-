```python

# ================================================================
# FACTORY PRODUCTION PLANNER (A7)
# B.Tech Mathematics Mini Project
#
# Mathematical Concepts:
#   1. Linear Equations
#   2. Consistency of a System
#   3. Rank Analysis
#   4. Practical Bottleneck Interpretation
#
# Environment:
#   Google Colab + Python
#   NumPy + SymPy + Pandas + Matplotlib
# ================================================================

# =========================
# 1. IMPORT LIBRARIES
# =========================

import numpy as np
import sympy as sp
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display, Markdown

print("=" * 70)
print("             FACTORY PRODUCTION PLANNER (A7)")
print("=" * 70)


# =========================
# 2. USER CONFIGURATION
# =========================
#
# ASSUMPTION:
# We assume the factory wants a production plan that uses
# exactly the available hours of all three machines.
#
# x1 = quantity of Product 1
# x2 = quantity of Product 2
# x3 = quantity of Product 3
#
# time_matrix:
# Rows    -> Machines
# Columns -> Products
#
# Example:
# Machine 1:
# Product 1 = 2 hours
# Product 2 = 1 hour
# Product 3 = 3 hours

products = ["Product 1", "Product 2", "Product 3"]

machines = ["Machine 1", "Machine 2", "Machine 3"]

# Time required by each product on each machine
time_matrix = np.array([
    [2, 1, 3],   # Machine 1
    [1, 2, 2],   # Machine 2
    [3, 2, 1]    # Machine 3
], dtype=float)

# Available hours for each machine
available_hours = np.array([
    100,   # Machine 1
    80,    # Machine 2
    90     # Machine 3
], dtype=float)


# =========================
# 3. DISPLAY INPUT DATA
# =========================

print("\nINPUT DATA")
print("-" * 70)

time_df = pd.DataFrame(
    time_matrix,
    index=machines,
    columns=products
)

print("\nTime required by each product on each machine (hours):")
display(time_df)

hours_df = pd.DataFrame({
    "Machine": machines,
    "Available Hours": available_hours
})

print("Available machine hours:")
display(hours_df)


# =========================
# 4. CONSTRUCT MATHEMATICAL MODEL
# =========================

print("\nMATHEMATICAL MODEL")
print("-" * 70)

# Symbolic variables
x1, x2, x3 = sp.symbols("x1 x2 x3")

variables = [x1, x2, x3]

# Convert to SymPy matrices
A = sp.Matrix(time_matrix.astype(int))
B = sp.Matrix(available_hours.astype(int))

# Matrix equation:
#
#       A X = B
#
# [ 2  1  3 ] [x1]   [100]
# [ 1  2  2 ] [x2] = [ 80]
# [ 3  2  1 ] [x3]   [ 90]

print("\nMatrix A:")
display(A)

print("Production vector X:")
display(sp.Matrix(variables))

print("Machine-hours vector B:")
display(B)

print("\nLinear Equations:")

for i in range(3):

    equation = sp.Eq(
        sum(A[i, j] * variables[j] for j in range(3)),
        B[i]
    )

    print(f"{machines[i]} : {equation}")


# =========================
# 5. RANK ANALYSIS
# =========================

print("\nRANK AND CONSISTENCY ANALYSIS")
print("-" * 70)

# Augmented matrix [A | B]
augmented_matrix = A.row_join(B)

rank_A = A.rank()
rank_augmented = augmented_matrix.rank()

number_of_variables = len(variables)

print("\nAugmented Matrix [A | B]:")
display(augmented_matrix)

print(f"Rank(A)       = {rank_A}")
print(f"Rank([A | B]) = {rank_augmented}")
print(f"Number of variables = {number_of_variables}")


# =========================
# 6. CHECK CONSISTENCY
# =========================
#
# Mathematical rule:
#
# System AX = B is consistent if:
#
#       Rank(A) = Rank([A|B])
#
# If ranks are different:
#       System is inconsistent.
#
# If ranks are equal to number of variables:
#       Unique solution.
#
# If ranks are equal but smaller than number of variables:
#       Infinitely many solutions.

if rank_A != rank_augmented:

    consistency_status = "INCONSISTENT"

elif rank_A == number_of_variables:

    consistency_status = "CONSISTENT - UNIQUE SOLUTION"

else:

    consistency_status = "CONSISTENT - INFINITELY MANY SOLUTIONS"


print("\nSystem Status:")
print(consistency_status)


# =========================
# 7. SOLVE SYSTEM
# =========================

print("\nPRODUCTION SOLUTION")
print("-" * 70)

solution = sp.linsolve((A, B), variables)

print("SymPy Solution:")
print(solution)


# =========================
# 8. NUMERICAL SOLUTION
# =========================

if rank_A == rank_augmented and rank_A == number_of_variables:

    solution_tuple = list(solution)[0]

    production_quantities = np.array(
        [float(value) for value in solution_tuple]
    )

    print("\nCalculated Production Quantities:")

    for product, quantity in zip(
        products,
        production_quantities
    ):

        print(
            f"{product:<15} : {quantity:.2f} units"
        )

else:

    # If there is no unique solution, use least-squares
    # only for practical bottleneck estimation.

    production_quantities, residuals, _, _ = np.linalg.lstsq(
        time_matrix,
        available_hours,
        rcond=None
    )

    print("\nNo unique exact production plan exists.")
    print("Least-squares estimate is shown for practical analysis.")

    for product, quantity in zip(
        products,
        production_quantities
    ):

        print(
            f"{product:<15} : {quantity:.2f} units"
        )


# =========================
# 9. VERIFY SOLUTION
# =========================

print("\nSOLUTION VERIFICATION")
print("-" * 70)

used_hours = time_matrix @ production_quantities

verification_df = pd.DataFrame({

    "Machine": machines,

    "Available Hours": available_hours,

    "Calculated Used Hours": used_hours,

    "Difference": used_hours - available_hours

})

display(verification_df)


# =========================
# 10. BOTTLENECK ANALYSIS
# =========================

print("\nBOTTLENECK ANALYSIS")
print("-" * 70)

absolute_difference = np.abs(
    used_hours - available_hours
)

bottleneck_index = np.argmax(
    absolute_difference
)

bottleneck_machine = machines[
    bottleneck_index
]

bottleneck_df = pd.DataFrame({

    "Machine": machines,

    "Available Hours": available_hours,

    "Estimated Used Hours": used_hours,

    "Absolute Difference": absolute_difference

})

display(bottleneck_df)


if rank_A != rank_augmented:

    print(
        f"Potential bottleneck: {bottleneck_machine}"
    )

    print(
        "Reason: This machine has the largest mismatch "
        "in the approximate production plan."
    )

else:

    print(
        "The system is mathematically consistent."
    )

    print(
        "Therefore, no bottleneck is identified from "
        "system inconsistency."
    )


# =========================
# 11. FINAL PRODUCTION TABLE
# =========================

print("\nFINAL PRODUCTION PLAN")
print("-" * 70)

production_df = pd.DataFrame({

    "Product": products,

    "Production Quantity": np.round(
        production_quantities,
        2
    )

})

display(production_df)


# =========================
# 12. MACHINE UTILIZATION
# =========================

utilization = (
    used_hours / available_hours
) * 100

utilization_df = pd.DataFrame({

    "Machine": machines,

    "Available Hours": available_hours,

    "Used Hours": np.round(
        used_hours,
        2
    ),

    "Utilization (%)": np.round(
        utilization,
        2
    )

})

print("\nMACHINE UTILIZATION")
print("-" * 70)

display(utilization_df)


# =========================
# 13. CHART 1
# AVAILABLE VS USED HOURS
# =========================

plt.figure(figsize=(10, 6))

x = np.arange(len(machines))

width = 0.35

plt.bar(
    x - width / 2,
    available_hours,
    width,
    label="Available Hours"
)

plt.bar(
    x + width / 2,
    used_hours,
    width,
    label="Used / Estimated Hours"
)

plt.xticks(
    x,
    machines
)

plt.xlabel("Machines")

plt.ylabel("Hours")

plt.title(
    "Factory Machine Capacity Analysis"
)

plt.legend()

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.show()


# =========================
# 14. CHART 2
# PRODUCTION QUANTITIES
# =========================

plt.figure(figsize=(9, 5))

plt.bar(
    products,
    production_quantities
)

plt.xlabel(
    "Products"
)

plt.ylabel(
    "Production Quantity"
)

plt.title(
    "Planned Production Quantities"
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.show()


# =========================
# 15. CHART 3
# MACHINE UTILIZATION
# =========================

plt.figure(figsize=(9, 5))

plt.bar(
    machines,
    utilization
)

plt.axhline(
    100,
    linestyle="--",
    label="100% Capacity"
)

plt.xlabel(
    "Machines"
)

plt.ylabel(
    "Utilization (%)"
)

plt.title(
    "Machine Utilization"
)

plt.legend()

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.show()


# =========================
# 16. BENCHMARK TEST
# =========================
#
# This benchmark uses the same realistic dataset.
# The test verifies that:
#
#   A × X = B
#
# for the calculated production quantities.

print("\nBENCHMARK TEST")
print("-" * 70)

benchmark_A = sp.Matrix([
    [2, 1, 3],
    [1, 2, 2],
    [3, 2, 1]
])

benchmark_B = sp.Matrix([
    100,
    80,
    90
])

benchmark_variables = sp.symbols(
    "x1 x2 x3"
)

benchmark_solution = sp.linsolve(
    (
        benchmark_A,
        benchmark_B
    ),
    benchmark_variables
)

benchmark_values = list(
    benchmark_solution
)[0]

benchmark_result = (
    benchmark_A *
    sp.Matrix(benchmark_values)
)

print("Benchmark solution:")

print(
    benchmark_solution
)

print("\nA × X:")

display(
    benchmark_result
)

print("Expected B:")

display(
    benchmark_B
)

if benchmark_result == benchmark_B:

    print(
        "✓ BENCHMARK TEST PASSED"
    )

else:

    print(
        "✗ BENCHMARK TEST FAILED"
    )


# =========================
# 17. FINAL AUTOMATIC REPORT
# =========================

print("\n")
print("=" * 70)
print("             FINAL PROJECT REPORT")
print("=" * 70)

print("\nMathematical Model:")
print("    A × X = B")

print("\nConsistency Result:")
print(
    f"    Rank(A) = {rank_A}"
)

print(
    f"    Rank([A|B]) = {rank_augmented}"
)

print(
    f"    Status = {consistency_status}"
)


if rank_A == rank_augmented and rank_A == number_of_variables:

    print("\nProduction Result:")

    for product, quantity in zip(
        products,
        production_quantities
    ):

        print(
            f"    {product}: {quantity:.2f} units"
        )

    print("\nPractical Interpretation:")

    print(
        "    A unique production plan exists."
    )

    print(
        "    All three machine-hour equations "
        "can be satisfied simultaneously."
    )

    print(
        "    No inconsistency-based bottleneck "
        "was detected."
    )


elif rank_A == rank_augmented:

    print("\nProduction Result:")

    print(
        "    The system is consistent, but there "
        "are infinitely many possible solutions."
    )

    print("\nPractical Interpretation:")

    print(
        "    The available equations are not "
        "sufficient to determine one unique "
        "production plan."
    )


else:

    print("\nProduction Result:")

    print(
        "    No exact production plan can satisfy "
        "all machine-hour equations simultaneously."
    )

    print("\nPractical Interpretation:")

    print(
        f"    Potential bottleneck: {bottleneck_machine}"
    )

    print(
        "    This machine shows the largest "
        "mismatch in the approximate plan."
    )


print("\n")
print("=" * 70)
print("              PROJECT COMPLETED")
print("=" * 70)
```
